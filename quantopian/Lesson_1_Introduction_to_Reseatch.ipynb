## **Getting Real Pricing Data**

Randomly sampled data can be great for testing ideas, but let's get some real data. We can use `get_pricing` to do that. You can use the `?` syntax as discussed above to get more information on `get_princing`'s arguments.

```python
import yfinance as yf

data = yf.download('MSFT', start='2012-1-1', end='2015-6-2', auto_adjust=False, actions=False)
```

Our data is now a datafram. You can see the datetime index and the colunms with different pricing data.

```python
data
```

```python
data['price'] = (data['Close'] + data['Open']) / 2
```

This is a pandas dataframe, so we can index in to just get price like this.

```python
X = data['price']
```

Because there is now also date information in our data, we provide two series to `.plot`. `x.index` gives us the datetime index, and `x.values` gives us the pricing values. These are used as the X and Y coordinates to make a graph.

```python
plt.plot(X.index, X.values)
plt.ylabel('Price')
plt.legend({'MSFT Returns'});
```

We can get statistics agains on real data

```python
np.mean(X)
```

```python
np.std(X)
```

## **Getting Returns from Prices**

We can use th `pct_change` function to get returns. Notice how we drop the first element after doing this, as if will be `Nan` (nothing => something results in a `Nan` percent change).

```python
R = X.pct_change()[1:]
```

We can plot the returns distribution as a histogram.

```python
plt.hist(R, bins = 20)
plt.xlabel('Returns')
plt.ylabel('Frequency')
plt.legend({'MSFT Returns'});
```

Get statistics again

```python
np.mean(R)
```

```python
np.std(R)
```

Now let's go backwards and generate data out of a normal distributions using the statistics we estimated from Microsoft's returns. We'll see that we have good reason to suspect Microsoft's returns may not be normal, as the resulting normal distribution looks far different.

```python
plt.hist(np.random.normal(np.mean(R), np.std(R), 10000), bins = 20)
plt.xlabel('Return')
plt.ylabel('Frquency')
plt.legend({'Normally Distributed returns'});
```

## **Generating a Moving Average**

`pandas` has some nice tools to allow us to generate rolling statistics. Here's an example. Notice how there's no moving average for the first 60 days, as we don't have60 days of data on which to generate the statistic.

```python
# Take the average of the last 60 days at each timepoint.
MAVG = X.rolling(window=60).mean()
plt.plot(X.index, X.values)
plt.plot(MAVG.index, MAVG.values)
plt.ylabel('Price')
plt.legend(['MSFT', '60-day MAVG']);
```
